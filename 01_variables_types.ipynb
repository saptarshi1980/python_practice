## 1. Values and types

Every value has a type. Common built-in types include `str` (text), `int` (whole numbers), `float` (decimal numbers), and `bool` (`True` or `False`). Python determines a value's type while the program runs.

````python
# Four values of different built-in types.
employee_name = "Manoj Kumar Khatua"       # str: text
employee_number = 11947            # int: whole number
monthly_basic_pay = 112_600.00     # float: decimal number, underscore in value 
is_active = True                  # bool: true/false value

print(employee_name)
print(employee_number)
print(monthly_basic_pay)
print(is_active)
````

## 2. Variables are names for values

In `employee_name = "Manoj Kumar Khatua"`, `=` assigns the value on the right to the name on the left. It does not mean “is equal to” in the comparison sense; comparisons use `==`.

Names are case-sensitive. Prefer descriptive `snake_case` names such as `monthly_basic_pay`. A name starts with a letter or underscore and cannot contain spaces.

````python
# type(...) returns a value's type. __name__ displays a short type name.
print(type(employee_name).__name__)
print(type(employee_number).__name__)
print(type(monthly_basic_pay).__name__)
print(type(is_active).__name__)
````

## 3. Arithmetic and assignment

Arithmetic creates a result. The calculation below stores that result under `annual_basic_pay`; `monthly_basic_pay` remains unchanged.

````python
annual_basic_pay = monthly_basic_pay * 12
print(annual_basic_pay)
print(monthly_basic_pay)  # Calculating annual pay does not change monthly basic pay.
````

## 4. Convert values before calculating

Text that looks like a number is still text. Convert it with `int(...)` for a whole number or `float(...)` for a decimal number before arithmetic. Conversion succeeds only when the text is a valid number.

````python
monthly_deduction_text = "25000"
monthly_deduction = int(monthly_deduction_text)
net_basic_pay = monthly_basic_pay - monthly_deduction
print(net_basic_pay)
print(type(monthly_deduction).__name__)
````

## 5. Format output with f-strings

An f-string starts with `f` and lets you put expressions inside `{}`. The format specifier `:,.2f` displays commas and two decimal places.

````python
print(f"{employee_name} earns {monthly_basic_pay:,.2f} per month.")
print(f"Net basic pay after deduction: {net_basic_pay:,.2f}")
````

## 6. Missing or unknown values: `None`

Use `None` for an absent or not-yet-known value. The string `"None"` is ordinary text and a different value. Check with `is None` or `is not None`.

````python
next_increment_date = None
print(next_increment_date is None)
print(next_increment_date is not None)
print(type("None").__name__)
````

## 7. Salary demo: allowances, deductions, and take-home pay

Run the earlier cells first to define `monthly_basic_pay` and `employee_name`. All amounts are in rupees.

Exercise assumptions:
- DA: 48% of monthly basic pay.
- HRA: 16% of monthly basic pay, with a monthly maximum of 16,000.
- Electricity allowance: an editable sample of 1,000 per month.
- PF: 12% of monthly basic pay plus DA, calculated as `(monthly_basic_pay + monthly_da) * pf_rate`.
- Annual taxable income: the portion of annual gross pay above 12,00,000. PF is a separate salary deduction.
- Estimated annual tax: `max(0, annual_gross_pay - 1_200_000) * 0.20`. Divide annual tax by 12 for the monthly deduction. Tax is zero when annual gross pay is at or below 12,00,000.

This is a simplified Python exercise, not an actual tax calculation. Slabs, exemptions, rebates, cess, and other adjustments are excluded. Salary is assumed constant for 12 months.

Percentages are decimal fractions: `48 / 100` is `0.48`. `min(...)` enforces the HRA cap; `max(0, ...)` prevents negative taxable income. Follow the numbered steps below.


````python
# Step 1: Define rates and an editable electricity allowance.
da_rate = 48 / 100
hra_rate = 16 / 100
monthly_hra_limit = 16_000.00
monthly_electricity_allowance = 1_000.00
pf_rate = 12 / 100
income_tax_rate = 20 / 100
annual_tax_free_threshold = 1_200_000.00
months_in_year = 12

# Step 2: Calculate each income head and gross monthly pay.
monthly_da = monthly_basic_pay * da_rate
monthly_hra = min(monthly_basic_pay * hra_rate, monthly_hra_limit)
monthly_gross_pay = (
    monthly_basic_pay + monthly_da + monthly_hra + monthly_electricity_allowance
)

# Step 3: Annualize income and PF. Tax only gross income above the threshold.
# Parentheses add basic pay and DA before applying the PF percentage.
monthly_pf = (monthly_basic_pay + monthly_da) * pf_rate
annual_gross_pay = monthly_gross_pay * months_in_year
annual_pf = monthly_pf * months_in_year
annual_taxable_income = max(0, annual_gross_pay - annual_tax_free_threshold)
annual_income_tax = annual_taxable_income * income_tax_rate
monthly_income_tax = annual_income_tax / months_in_year

# Step 4: Subtract deductions to calculate take-home pay.
monthly_total_deductions = monthly_pf + monthly_income_tax
monthly_net_pay = monthly_gross_pay - monthly_total_deductions
annual_net_pay = monthly_net_pay * months_in_year

# Step 5: Display amounts with commas and two decimal places.
print(f"Employee: {employee_name}")
print(f"Basic monthly pay: {monthly_basic_pay:,.2f}")
print(f"DA (48%): {monthly_da:,.2f}")
print(f"HRA (16%, capped at 16,000): {monthly_hra:,.2f}")
print(f"Electricity allowance: {monthly_electricity_allowance:,.2f}")
print(f"Gross monthly pay: {monthly_gross_pay:,.2f}")
print(f"Monthly PF (12% of basic + DA): {monthly_pf:,.2f}")
print(f"Annual gross pay: {annual_gross_pay:,.2f}")
print(f"Annual PF: {annual_pf:,.2f}")
print(f"Annual tax-free threshold: {annual_tax_free_threshold:,.2f}")
print(f"Annual income above threshold (exercise): {annual_taxable_income:,.2f}")
print(f"Estimated annual tax (20%): {annual_income_tax:,.2f}")
print(f"Estimated monthly tax: {monthly_income_tax:,.2f}")
print(f"Total monthly deductions: {monthly_total_deductions:,.2f}")
print(f"Monthly take-home pay: {monthly_net_pay:,.2f}")
print(f"Annual take-home pay: {annual_net_pay:,.2f}")
````
