## 1. Refresh arithmetic operators

You have used `+`, `-`, `*`, and `/`. Three more operators are useful:

| Operator | Meaning | Example |
| --- | --- | --- |
| `//` | Floor division: round the quotient down | `17 // 5` gives `3` |
| `%` | Remainder | `17 % 5` gives `2` |
| `**` | Raise to a power | `2 ** 3` gives `8` |

For nonnegative whole numbers, floor division counts complete groups and the remainder counts what is left. Floor division rounds toward negative infinity: `-17 // 5` is `-4`.

Multiplication and division happen before addition and subtraction. Parentheses make the intended order explicit.

````python
training_minutes = 137
print("Complete hours:", training_minutes // 60)
print("Remaining minutes:", training_minutes % 60)
print("Two cubed:", 2 ** 3)
print("Ordinary division:", 17 / 5)
print("Without parentheses:", 10 + 5 * 2)
print("With parentheses:", (10 + 5) * 2)
````

## 2. Compare values

Comparisons produce a `bool`: either `True` or `False`.

| Operator | Meaning |
| --- | --- |
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

`=` assigns a value; `==` compares values. Equality with strings is case-sensitive. Convert numeric text before comparing it with a number.

````python
monthly_basic_pay = 75_000.00
is_active = True
employee_name = "Pijush Kanti Pal"

print(monthly_basic_pay == 75_000)
print(monthly_basic_pay != 75_000)
print(monthly_basic_pay > 75_000)
print(monthly_basic_pay < 80_000)
print(monthly_basic_pay >= 75_000)
print(monthly_basic_pay <= 75_000)
print("Pijush" == "pijush")
is_above_limit = monthly_basic_pay > 70_000
print(is_above_limit, type(is_above_limit).__name__)
````

## 3. Make a decision with `if`

An `if` statement runs its indented block only when the condition is true. End the condition with a colon (`:`) and indent the block with four spaces.

The final print below is outside the block, so it always runs. Try changing `is_active` to `False`.

````python
is_active = True
if is_active:
    print("Employee account is active.")
    print("Payroll details can be reviewed.")
print("Status check finished.")
````

## 4. Choose between two paths with `if` and `else`

An `else` block runs when the `if` condition is false. `else` has no condition. Both branches below assign `annual_taxable_income`, so it always exists afterward.

Annual gross equal to the threshold has zero tax under our worksheet formula. Change annual gross to `1_200_000.00` and then `1_100_000.00` to check.

````python
annual_gross_pay = 1_494_000.00
annual_tax_free_threshold = 1_200_000.00
income_tax_rate = 20 / 100

if annual_gross_pay > annual_tax_free_threshold:
    annual_taxable_income = annual_gross_pay - annual_tax_free_threshold
else:
    annual_taxable_income = 0.00

annual_income_tax = annual_taxable_income * income_tax_rate
monthly_income_tax = annual_income_tax / 12
print(f"Annual income above threshold: {annual_taxable_income:,.2f}")
print(f"Annual tax: {annual_income_tax:,.2f}")
print(f"Monthly tax: {monthly_income_tax:,.2f}")
````

## 5. Choose among several paths with `elif`

Python checks conditions from top to bottom and runs only the first matching branch. Put the highest score threshold first here; otherwise a score of 95 could match the 60-point branch too soon.

These bonus rates are new exercise rules: scores of 90-100 earn 10%, scores of 75-89 earn 5%, scores of 60-74 earn 2%, and scores below 60 earn no bonus. For this first example, assume the score is a valid whole number from 0 to 100.

````python
performance_score = 82
bonus_basic_pay = 50_000.00

if performance_score >= 90:
    bonus_rate = 0.10
elif performance_score >= 75:
    bonus_rate = 0.05
elif performance_score >= 60:
    bonus_rate = 0.02
else:
    bonus_rate = 0.00

bonus_amount = bonus_basic_pay * bonus_rate
print(f"Bonus rate: {bonus_rate:.0%}")
print(f"Bonus amount: {bonus_amount:,.2f}")
````

## 6. Combine conditions with `and`, `or`, and `not`

For boolean conditions:

| Operator | Result is true when... |
| --- | --- |
| `and` | Both conditions are true |
| `or` | At least one condition is true |
| `not` | The condition is false |

Use parentheses when combining several rules so your intent is clear. Prefer `if is_active:` to `if is_active == True:`.

````python
is_active = True
completed_service_months = 14
has_travel_receipts = False
has_manager_approval = True

is_bonus_eligible = is_active and completed_service_months >= 12
can_claim_travel = is_active and (has_travel_receipts or has_manager_approval)

print("Bonus eligible:", is_bonus_eligible)
print("Travel claim allowed:", can_claim_travel)
print("Account inactive:", not is_active)

if is_bonus_eligible:
    print("Employee meets the bonus eligibility rules.")
else:
    print("Employee does not meet the bonus eligibility rules.")
````

## 7. Check ranges and missing values

A chained comparison such as `0 <= score <= 100` checks both bounds, including the endpoints. Check invalid inputs before making a business decision.

Use `is None` for a missing value. The number zero can be a valid known value, so a check such as `if score:` cannot distinguish a missing score from a zero score. A guarded condition can use `and`: Python skips its right side when the left side is false. This is called short-circuit evaluation.

````python
performance_score = None

if performance_score is None:
    print("Enter a performance score before calculating a bonus.")
elif not (0 <= performance_score <= 100):
    print("Score must be between 0 and 100.")
else:
    print("Score is valid.")

# The comparison is skipped because the score is None.
has_passing_score = performance_score is not None and performance_score >= 60
print("Has a passing score:", has_passing_score)
````

## 8. Apply conditions to a complete salary calculation

This example uses the same salary rules as Lesson 1. It replaces `min(...)` for HRA and `max(...)` for taxable income with explicit decisions.

Inputs must be nonnegative for this exercise. The validation branch prevents calculation when an input is negative. Electricity allowance is a fixed monthly input. PF is calculated on **basic plus DA**, while tax is based on **annual gross above the threshold**.

````python
# Keep the employee_name defined earlier in this notebook.
monthly_basic_pay = 75_000.00
monthly_electricity_allowance = 1_500.00
da_rate = 0.48
hra_rate = 0.16
monthly_hra_limit = 16_000.00
pf_rate = 0.12
annual_tax_free_threshold = 1_200_000.00
income_tax_rate = 0.20
months_in_year = 12

if monthly_basic_pay < 0 or monthly_electricity_allowance < 0:
    print("Basic pay and electricity allowance must be nonnegative.")
else:
    monthly_da = monthly_basic_pay * da_rate
    calculated_hra = monthly_basic_pay * hra_rate
    if calculated_hra > monthly_hra_limit:
        monthly_hra = monthly_hra_limit
    else:
        monthly_hra = calculated_hra

    monthly_gross_pay = (
        monthly_basic_pay + monthly_da + monthly_hra + monthly_electricity_allowance
    )
    monthly_pf = (monthly_basic_pay + monthly_da) * pf_rate
    annual_gross_pay = monthly_gross_pay * months_in_year

    if annual_gross_pay > annual_tax_free_threshold:
        annual_taxable_income = annual_gross_pay - annual_tax_free_threshold
    else:
        annual_taxable_income = 0.00

    annual_income_tax = annual_taxable_income * income_tax_rate
    monthly_income_tax = annual_income_tax / months_in_year
    monthly_total_deductions = monthly_pf + monthly_income_tax
    monthly_net_pay = monthly_gross_pay - monthly_total_deductions
    annual_net_pay = monthly_net_pay * months_in_year

    print(f"Employee: {employee_name}")
    print(f"Basic pay: {monthly_basic_pay:,.2f}")
    print(f"DA: {monthly_da:,.2f}")
    print(f"HRA: {monthly_hra:,.2f}")
    print(f"Electricity allowance: {monthly_electricity_allowance:,.2f}")
    print(f"Monthly gross: {monthly_gross_pay:,.2f}")
    print(f"Monthly PF (basic + DA): {monthly_pf:,.2f}")
    print(f"Annual gross: {annual_gross_pay:,.2f}")
    print(f"Annual income above threshold: {annual_taxable_income:,.2f}")
    print(f"Annual tax: {annual_income_tax:,.2f}")
    print(f"Monthly tax: {monthly_income_tax:,.2f}")
    print(f"Monthly deductions: {monthly_total_deductions:,.2f}")
    print(f"Monthly take-home: {monthly_net_pay:,.2f}")
    print(f"Annual take-home: {annual_net_pay:,.2f}")
````

**Check:** monthly gross = `124,500.00`; PF = `13,320.00`; annual tax = `58,800.00`; monthly tax = `4,900.00`; monthly take-home = `106,280.00`.

## 9. Common mistakes to recognize

The incorrect examples below are commented out so **Run All** still works. To explore an error, uncomment just one example, run the cell, read the message, and then comment it again.

- Assignment (`=`) cannot replace equality (`==`) in an ordinary `if` condition.
- A missing colon causes a syntax error.
- A block needs consistent indentation.
- `"False"` is nonempty text, so it is truthy; use the boolean `False` for a false flag.
- Ordered comparisons such as `"75000" > 70000` require converting the text first.
- Use `==` for value equality; reserve `is None` for checking missing values.

````python
# SyntaxError: assignment used instead of comparison.
# if monthly_basic_pay = 75000:
#     print("Equal")

# SyntaxError: missing colon.
# if monthly_basic_pay > 0
#     print("Positive")

# IndentationError: no indented block.
# if monthly_basic_pay > 0:
# print("Positive")

# TypeError: ordered comparison between text and a number.
# print("75000" > 70000)

# Correct comparison after conversion:
print(float("75000") > 70000)
print("A nonempty string is truthy:", bool("False"))
print("A boolean false value:", bool(False))
````
