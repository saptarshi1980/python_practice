## 1. Create strings and include special characters

A `str` is a sequence of Unicode characters. Single and double quotes both create strings.
Choose quotes that make text easy to read. Triple quotes allow multiple lines.
`\n` represents a newline and `\t` a tab. A raw string such as `r"C:\notes"` treats
backslashes literally, which is useful for Windows paths. A raw string cannot end with a single backslash.

```python
employee_name = "Asha Sen"
department = 'Finance'
message = "It's payday!"
note = """Review your payslip.
Check the deductions."""
print(employee_name, department)
print(message)
print(note)
print("Name\tDepartment\nAsha\tFinance")
print(r"C:\notes\salary.txt")
```

## 2. Length, indexing, and slicing

`len(text)` counts characters. Indexes start at zero: `text[0]` is the first character.
`text[-1]` is the last. `text[start:stop]` includes the start but excludes the stop.
Omit a bound to use the beginning or end. `text[::2]` takes every second character.

An individual index outside the string raises `IndexError`. Slices safely stop at the end.
An empty string has length zero, so check it before using `[0]` or `[-1]`.

```python
employee_code = "EMP-1042"
print("Length:", len(employee_code))
print("First:", employee_code[0])
print("Last:", employee_code[-1])
print("Prefix:", employee_code[:3])
print("Number text:", employee_code[4:])
print("Every second character:", employee_code[::2])
print("Long slice:", employee_code[:100])
empty_text = ""
if empty_text:
    print(empty_text[0])
else:
    print("No first character: text is empty.")
```

## 3. Strings are immutable

**Immutable** means you cannot replace a character inside an existing string.
Methods such as `upper()` and `replace()` produce new strings; they do not change the original.
You can assign the result back to a variable. `+` joins strings and `*` repeats a string.

```python
employee_name = "Asha Sen"
uppercase_name = employee_name.upper()
print("Original:", employee_name)
print("New:", uppercase_name)
corrected_name = employee_name.replace("Sen", "Das")
print(corrected_name)
print("Employee: " + employee_name)
print("-" * 25)
# Uncomment to see TypeError, then comment it again:
# employee_name[0] = "a"
```

## 4. Clean and compare text

`strip()` removes whitespace from both ends, not the middle. `lower()` and `upper()` change case.
`casefold()` is useful for comparisons that ignore case. `title()` is a simple display transformation;
it is not a reliable way to correct every person's name. `replace(old, new)` replaces matching text.

```python
raw_department = "  Finance  "
clean_department = raw_department.strip()
print(f"Before: [{raw_department}]; after: [{clean_department}]")
print(clean_department.lower())
print(clean_department.upper())
print("Matches finance:", clean_department.casefold() == "finance")
print("monthly salary report".title())
print("EMP/1042".replace("/", "-"))
```

## 5. Search text and split simple fields

`in` tests whether text occurs anywhere. `startswith(...)` and `endswith(...)` test the edges.
These checks are case-sensitive unless you normalize the text first.
`split(separator)` returns a list of pieces. `separator.join(pieces)` combines strings.
For now, print the whole list; Lesson 4 explains how lists work.

These examples handle simple delimited text. Real CSV files with quoted commas need a CSV reader.

```python
filename = "salary_report.txt"
print("salary" in filename)
print(filename.startswith("salary_"))
print(filename.endswith(".txt"))
print("SALARY" in filename)
record = "Asha Sen,Finance,1042"
fields = record.split(",")
print(fields)
print(" | ".join(fields))
print("  Asha   Sen  ".split())
```

## 6. Convert numeric text and format a short summary

Clean numeric text before converting it. `float(...)` does not accept currency symbols or commas:
remove those only when the expected input format is known. Arbitrary text such as `"unknown"`
still cannot be converted; handling conversion exceptions comes later.

An f-string inserts values. `:,.2f` displays money with commas and two decimal places;
`:.0%` displays a decimal rate as a percentage. Formatting creates text, not a new numeric amount.

```python
employee_name = "  Arun Paul  ".strip()
basic_text = " 75,000.00 "
basic_pay = float(basic_text.strip().replace(",", ""))
da_rate = 0.48
monthly_da = basic_pay * da_rate
summary = f"{employee_name}: basic {basic_pay:,.2f}, DA ({da_rate:.0%}) {monthly_da:,.2f}"
print(summary)
print(type(basic_pay).__name__, type(summary).__name__)
# This would fail: float("75,000.00")
```
