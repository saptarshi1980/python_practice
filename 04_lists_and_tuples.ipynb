## 1. Store ordered values in a list

A list uses square brackets and keeps insertion order. It can contain duplicates and mixed types,
although one consistent kind of item often makes calculations easier. `len(...)` counts items.
Indexing and slicing follow the same rules as strings: zero-based indexes, negative indexes,
and an excluded slice stop. Invalid individual indexes raise `IndexError`.

```python
employee_names = ["Asha", "Arun", "Neha", "Asha"]
print(employee_names)
print("Count:", len(employee_names))
print("First:", employee_names[0])
print("Last:", employee_names[-1])
print("First two:", employee_names[:2])
print("Arun present:", "Arun" in employee_names)
print("Missing present:", "Rina" in employee_names)
```

## 2. Lists are mutable

You can change an item with assignment. `append(value)` adds one item at the end.
`extend(values)` adds each item from another collection. `insert(index, value)` inserts an item.
These methods change the list and return `None`; do not assign their result back to your list.

```python
names = ["Asha", "Arun"]
names[1] = "Neha"
names.append("Rina")
print("After append:", names)
names.extend(["Manoj", "Dev"])
names.insert(1, "Isha")
print("After extend and insert:", names)
append_result = names.append("Amit")
print("Method returned:", append_result)
print("Updated list:", names)
```

## 3. Remove items carefully

`remove(value)` removes the first matching value and raises `ValueError` if no match exists.
`pop()` removes and returns the last item; `pop(index)` uses a position. Popping from an empty
list raises `IndexError`. `del names[index]` removes an item without returning it.
Use membership and nonempty checks when a value or item might be absent.

```python
names = ["Asha", "Arun", "Asha", "Neha"]
names.remove("Asha")
print("First match removed:", names)
last_name = names.pop()
print("Removed last:", last_name)
del names[0]
print("Remaining:", names)
if "Rina" in names:
    names.remove("Rina")
else:
    print("Rina is not in the list.")
empty_names = []
if empty_names:
    print(empty_names.pop())
else:
    print("Nothing to pop.")
```

## 4. Summarize and sort numeric values

`sum(...)`, `min(...)`, and `max(...)` summarize compatible numeric values. `sum([])` is zero,
but `min([])` and `max([])` raise `ValueError`. Guard averages against a zero count.

`sorted(values)` returns a new list. `values.sort()` changes an existing list and returns `None`.
`reverse=True` requests descending order. Avoid mixing numbers and strings in ordering operations.

```python
basic_salaries = [60_000.00, 40_000.00, 50_000.00]
print("Total:", sum(basic_salaries))
if basic_salaries:
    print("Lowest:", min(basic_salaries))
    print("Highest:", max(basic_salaries))
    print("Average:", sum(basic_salaries) / len(basic_salaries))
ascending = sorted(basic_salaries)
print("New sorted list:", ascending)
print("Original:", basic_salaries)
basic_salaries.sort(reverse=True)
print("Original after sorting:", basic_salaries)
```

## 5. Assignment is not a copy

Assigning a list to a second name makes both names refer to the same list. Updating through
either name affects that shared list. `.copy()` makes a separate outer list.
It is a shallow copy: nested lists would still be shared. Here our items are simple immutable numbers.

```python
original_pay = [40_000.00, 50_000.00]
shared_pay = original_pay
separate_pay = original_pay.copy()
shared_pay[0] = 42_000.00
print("Original:", original_pay)
print("Shared:", shared_pay)
print("Separate copy:", separate_pay)
```

## 6. Tuples: ordered values with fixed items

A tuple is an immutable sequence: you cannot replace, append, or remove its items.
Commas create a tuple; parentheses usually make it easier to read. A one-item tuple needs a
trailing comma: `(1042,)`. `(1042)` is just an integer in parentheses. `()` is an empty tuple.

Tuples support length, indexing, slicing, duplicates, and membership. They can hold mutable
objects, so immutability applies to the tuple's item references rather than everything inside it.
For now, use simple immutable fields such as names and numbers.

```python
employee = (1042, "Asha Sen", 52_500.00)
print(employee)
print("Name:", employee[1])
print("First two fields:", employee[:2])
print("Length:", len(employee))
print("Name present:", "Asha Sen" in employee)
print(type((1042,)).__name__)
print(type((1042)).__name__)
# Uncomment to see TypeError, then comment it again:
# employee[2] = 55_000.00
```

## 7. Pack, unpack, and convert

Packing groups values into a tuple. Unpacking assigns its items to separate names.
For simple unpacking, the number of names must match the number of items; otherwise Python
raises `ValueError`. Use `list(...)` or `tuple(...)` to create a collection of another type.

Choose a list for a collection you intend to update, such as employees awaiting review.
Choose a tuple for a fixed grouping, such as a pair of coordinates or a simple employee record.

```python
employee_record = (2075, "Rina Das", 68_400.00)
employee_number, employee_name, basic_pay = employee_record
print(f"{employee_number}: {employee_name}, basic {basic_pay:,.2f}")
editable_record = list(employee_record)
editable_record[2] = 70_000.00
updated_record = tuple(editable_record)
print("Original tuple:", employee_record)
print("New tuple:", updated_record)
```
