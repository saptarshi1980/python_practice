## 1. Why use a loop?

A loop repeats a block of code. Instead of writing a reminder separately for every month, write the action once.

`for month in range(1, 4):` assigns each number to `month` in turn. The indented block runs for 1, 2, and 3. A single pass is called an **iteration**. The print outside the loop runs once, after all iterations.

```python
for month in range(1, 4):
    print(f"Month {month}: review your payslip")
print("All reminders printed.")
```

## 2. Understand range

| Expression | Values visited |
| --- | --- |
| `range(4)` | 0, 1, 2, 3 |
| `range(1, 4)` | 1, 2, 3 |
| `range(2, 9, 2)` | 2, 4, 6, 8 |
| `range(3, 0, -1)` | 3, 2, 1 |

The start is included and the stop is excluded. The default start is zero and the default step is one. The step cannot be zero. An empty range runs the body zero times.

```python
print("Starting at zero:")
for number in range(4):
    print(number)
print("Even numbers:")
for number in range(2, 9, 2):
    print(number)
print("Countdown:")
for number in range(3, 0, -1):
    print(number)
for number in range(2, 2):
    print("This does not run.")
```

## 3. Visit values in a list

A **list** holds several values in order, inside square brackets. A `for` loop visits each value directly. Duplicate values are visited too.

This revisits Lesson 4: the loop visits list values directly, without needing indexes.

```python
employee_names = ["Sujit", "Arun", "Ashok"]
for employee_name in employee_names:
    print(f"Welcome, {employee_name}!")

basic_pay_samples = [40_000.00, 50_000.00, 40_000.00]
for basic_pay in basic_pay_samples:
    monthly_da = basic_pay * 0.48
    print(f"Basic: {basic_pay:,.2f}; DA: {monthly_da:,.2f}")
```

## 4. Keep a running total

An **accumulator** holds a running total. Initialize it before the loop and add each amount inside the loop. A counter increases by one for each item.

For numbers, `total += amount` is a shorter way to write `total = total + amount`. Resetting the total inside the loop would lose earlier amounts.

```python
allowances = [1_000.00, 1_500.00, 2_000.00]
total_allowance = 0.00
allowance_count = 0
for allowance in allowances:
    total_allowance += allowance
    allowance_count += 1
    print(f"After item {allowance_count}: {total_allowance:,.2f}")
print(f"Total: {total_allowance:,.2f}")
if allowance_count > 0:
    print(f"Average: {total_allowance / allowance_count:,.2f}")
else:
    print("No allowances to average.")
```

**Check:** total = 4,500; count = 3; average = 1,500. Try an empty list `[]`: the total remains zero and the condition prevents division by zero.

A fixed monthly contribution can be accumulated in the same way. Notice which prints are inside and outside the loop.

```python
monthly_savings = 2_500.00
savings_total = 0.00
for month in range(1, 5):
    savings_total += monthly_savings
    print(f"Month {month}: {savings_total:,.2f}")
print(f"Four-month savings: {savings_total:,.2f}")
```

## 5. Combine loops and conditions

A condition inside the loop is checked for each value. Calculate HRA at 16% of basic, capped at 16,000. Assign HRA in both branches on every iteration so an earlier result is not reused.

```python
for basic_pay in [52_500.00, 100_000.00, 120_000.00]:
    calculated_hra = basic_pay * 0.16
    if calculated_hra > 16_000:
        monthly_hra = 16_000.00
        message = "capped"
    else:
        monthly_hra = calculated_hra
        message = "within limit"
    print(f"Basic: {basic_pay:,.2f}; HRA: {monthly_hra:,.2f} ({message})")
```

## 6. Repeat while a condition is true

Use `for` to visit values or count a known range. Use `while` when repetition depends on a condition. Python checks the condition before each iteration, so the body may run zero times.

Save a positive contribution each month until a target is reached. Update the savings inside the loop so it can finish. Without this update, the loop could run forever. Use Jupyter's **Interrupt Kernel** button if a loop runs unexpectedly, then fix the condition or update.

```python
savings_target = 10_000.00
monthly_contribution = 2_500.00
saved_amount = 0.00
months_needed = 0
if monthly_contribution <= 0:
    print("Monthly contribution must be positive.")
else:
    while saved_amount < savings_target:
        saved_amount += monthly_contribution
        months_needed += 1
        print(f"Month {months_needed}: {saved_amount:,.2f}")
    print(f"Target reached in {months_needed} months.")
```

**Check:** four months, 10,000 saved. Change the target to 11,000: five contributions give 12,500. Reaching a target can overshoot it. A target of zero needs zero iterations. A zero contribution must show the validation message.

## 7. Skip an item with continue

`continue` skips the rest of the current iteration and moves to the next item. Here, negative basic pay values are invalid and excluded from the total. Zero remains valid.

```python
valid_pay_total = 0.00
for basic_pay in [50_000.00, -500.00, 75_000.00]:
    if basic_pay < 0:
        print(f"Skipping invalid pay: {basic_pay:,.2f}")
        continue
    valid_pay_total += basic_pay
    print(f"Accepted: {basic_pay:,.2f}")
print(f"Total valid basic pay: {valid_pay_total:,.2f}")
```

## 8. Stop a loop with break

`break` exits the loop immediately. Code after the loop still runs. Find the first annual income above the worksheet tax threshold. A boolean flag lets us report when no match was found, including when the list is empty.

```python
annual_gross_samples = [900_000.00, 1_200_000.00, 1_500_000.00, 1_800_000.00]
found_match = False
for annual_gross in annual_gross_samples:
    if annual_gross > 1_200_000:
        print(f"First annual gross above threshold: {annual_gross:,.2f}")
        found_match = True
        break
if not found_match:
    print("No income exceeds the threshold.")
print("Search finished.")
```

## 9. Calculate several sample payslips

Each list item represents a separate employee's monthly basic pay. Electricity allowance is 1,000 per month for every sample. Use the same worksheet rules as the salary lessons:

- DA = 48% of basic.
- HRA = 16% of basic, capped at 16,000 monthly.
- PF = 12% of **basic plus DA**.
- Annual gross = monthly gross multiplied by 12.
- Annual tax = 20% of annual gross **above 12,00,000**; zero at or below the threshold.
- Monthly tax = annual tax divided by 12. PF remains a separate deduction.

These are simplified learning assumptions, not actual payroll or tax rules. Assume constant pay for 12 months per employee. Calculate each person's tax first, then add totals. The summary combines **one month's** deductions and take-home across employees.

```python
pay_samples = [52_500.00, 112_600.00, 120_000.00]
electricity_allowance = 1_000.00
employee_count = 0
total_monthly_pf = 0.00
total_monthly_tax = 0.00
total_monthly_net = 0.00

for basic_pay in pay_samples:
    if basic_pay < 0:
        print(f"Skipping invalid basic pay: {basic_pay:,.2f}")
        continue
    monthly_da = basic_pay * 0.48
    monthly_hra = min(basic_pay * 0.16, 16_000.00)
    monthly_gross = basic_pay + monthly_da + monthly_hra + electricity_allowance
    monthly_pf = (basic_pay + monthly_da) * 0.12
    annual_gross = monthly_gross * 12
    annual_taxable_income = max(0, annual_gross - 1_200_000.00)
    annual_tax = annual_taxable_income * 0.20
    monthly_tax = annual_tax / 12
    monthly_net = monthly_gross - monthly_pf - monthly_tax

    employee_count += 1
    total_monthly_pf += monthly_pf
    total_monthly_tax += monthly_tax
    total_monthly_net += monthly_net
    print(f"Sample {employee_count}; basic: {basic_pay:,.2f}")
    print(f"  Gross: {monthly_gross:,.2f}; PF: {monthly_pf:,.2f}")
    print(f"  Monthly tax: {monthly_tax:,.2f}; take-home: {monthly_net:,.2f}")

print(f"Valid employees: {employee_count}")
print(f"Combined monthly PF: {total_monthly_pf:,.2f}")
print(f"Combined monthly tax: {total_monthly_tax:,.2f}")
print(f"Combined monthly take-home: {total_monthly_net:,.2f}")
if employee_count > 0:
    print(f"Average monthly take-home: {total_monthly_net / employee_count:,.2f}")
else:
    print("No valid salaries to average.")
```

**Check:**

| Monthly basic | Monthly PF | Monthly tax | Monthly take-home |
| --- | ---: | ---: | ---: |
| 52,500.00 | 9,324.00 | 0.00 | 77,776.00 |
| 112,600.00 | 19,997.76 | 16,729.60 | 146,920.64 |
| 120,000.00 | 21,312.00 | 18,920.00 | 154,368.00 |

Combined monthly PF = **50,633.76**; tax = **35,649.60**; take-home = **379,064.64**. Average monthly take-home = **126,354.88**.

## 10. Common mistakes

- `range(1, 12)` stops at 11. Use `range(1, 13)` for all 12 months.
- Initialize totals before the loop, not inside it.
- Indent the statements that must repeat.
- Update the controlling variable inside a `while` loop.
- Reset totals when rerunning calculations.
- An empty collection never assigns the loop variable: do not rely on it after the loop.

Compare the incorrect and corrected totals below.

```python
for amount in [1, 2, 3]:
    incorrect_total = 0
    incorrect_total += amount
print("Incorrect total:", incorrect_total)

correct_total = 0
for amount in [1, 2, 3]:
    correct_total += amount
print("Correct total:", correct_total)
```
