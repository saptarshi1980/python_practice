## 1. Dictionaries: look up values by key

A dictionary maps unique keys to values using `{key: value}`. Use descriptive string keys
for employee fields. Unlike a list position, a key explains what a value represents.
Dictionaries preserve insertion order, but they are mappings rather than positional sequences.
Keys must be hashable; strings and numbers work, lists do not. Values can be any type.

```python
employee = {"number": 4050, "name": "Arun Paul", "basic_pay": 75_000.00, "active": True}
print(employee)
print(employee["name"])
print(employee["basic_pay"])
print("Field count:", len(employee))
print("name" in employee)
print("Arun Paul" in employee)  # Membership checks keys, not values.
```

## 2. Read missing fields safely

`record[key]` raises `KeyError` when the key is missing. `record.get(key)` returns `None`
for a missing key, or you can supply a default. `.get(...)` does not insert a key.
Use `key in record` to distinguish a missing key from a key whose value is `None`.

```python
employee = {"name": "Arun Paul", "increment_date": None}
print("Department:", employee.get("department", "Not recorded"))
print("Increment date:", employee.get("increment_date"))
print("Date field present:", "increment_date" in employee)
print("Department field present:", "department" in employee)
# Uncomment to see KeyError, then comment it again:
# print(employee["department"])
```

## 3. Add, update, remove, and copy fields

Assignment adds a new key or updates an existing one. A key appears only once; assigning it
again replaces its value. `update(...)` changes or adds several fields. `pop(key, default)`
removes a field and returns its value, or the default when absent. `del record[key]` removes
a required field and raises `KeyError` if absent.

Like lists, dictionaries are mutable. Assignment shares the dictionary; `.copy()` creates a
separate outer dictionary. Nested mutable values are still shared by this shallow copy.

```python
employee = {"name": "Arun Paul", "basic_pay": 75_000.00}
employee["department"] = "Finance"
employee["basic_pay"] = 78_000.00
employee.update({"active": True, "temporary_note": "Review"})
removed_note = employee.pop("temporary_note", None)
print("Removed note:", removed_note)
print("Missing field:", employee.pop("unknown", "Not found"))
copied_employee = employee.copy()
copied_employee["basic_pay"] = 80_000.00
print("Original basic:", employee["basic_pay"])
print("Copied basic:", copied_employee["basic_pay"])
print(employee)
```

## 4. Visit keys, values, and pairs

Looping over a dictionary directly visits keys. `.keys()`, `.values()`, and `.items()` provide
keys, values, or key-value pairs. Unpack each pair in an items loop, as you unpacked tuples.
Do not add or remove dictionary keys while iterating over that dictionary.

```python
income_heads = {"basic": 75_000.00, "da": 36_000.00, "hra": 12_000.00, "electricity": 1_500.00}
print("Keys:", list(income_heads.keys()))
print("Values:", list(income_heads.values()))
gross_pay = 0.00
for head, amount in income_heads.items():
    print(f"{head}: {amount:,.2f}")
    gross_pay += amount
print(f"Monthly gross: {gross_pay:,.2f}")
```

## 5. A list of employee dictionaries

A list can hold several dictionaries, one per employee. This combines ordered records with
named fields. Calculate each employee's deductions separately before totaling take-home pay.

Use our existing worksheet rules: DA 48%; HRA 16% capped at 16,000; electricity 1,000;
PF 12% of basic plus DA; annual tax 20% only on annual gross above 12,00,000, divided by
12 for monthly tax. These are simplified exercise assumptions, not actual payroll rules.
Assume nonnegative salary inputs and constant pay for 12 months.

```python
employees = [
    {"name": "Asha Sen", "basic_pay": 52_500.00},
    {"name": "Manoj", "basic_pay": 112_600.00},
]
combined_net = 0.00
for employee in employees:
    basic_pay = employee["basic_pay"]
    monthly_da = basic_pay * 0.48
    monthly_hra = min(basic_pay * 0.16, 16_000.00)
    monthly_gross = basic_pay + monthly_da + monthly_hra + 1_000.00
    monthly_pf = (basic_pay + monthly_da) * 0.12
    annual_gross = monthly_gross * 12
    annual_tax = max(0, annual_gross - 1_200_000.00) * 0.20
    monthly_tax = annual_tax / 12
    monthly_net = monthly_gross - monthly_pf - monthly_tax
    combined_net += monthly_net
    print(f"{employee['name']}: monthly take-home {monthly_net:,.2f}")
print(f"Combined monthly take-home: {combined_net:,.2f}")
```

**Check:** Asha = 77,776.00; Manoj = 146,920.64; combined monthly take-home = 224,696.64.

## 6. Sets: store unique values

A set contains unique, hashable values and has no guaranteed display or iteration order.
Use `set()` for an empty set: `{}` creates an empty dictionary. Sets do not support indexing
or slicing. `set(values)` removes duplicates but does not preserve the list's order.
Use `sorted(...)` when you want predictable display of compatible values.

```python
departments = ["Finance", "HR", "Finance", "IT"]
unique_departments = set(departments)
print("Unique department count:", len(unique_departments))
print("Sorted for display:", sorted(unique_departments))
print("Finance present:", "Finance" in unique_departments)
empty_set = set()
empty_dictionary = {}
print(type(empty_set).__name__, type(empty_dictionary).__name__)
# unique_departments[0] would raise TypeError.
```

## 7. Update sets and compare groups

`add(value)` adds one item. Adding an existing item has no effect. `discard(value)` removes
an item if present and does nothing if absent; `remove(value)` raises `KeyError` if absent.
These update methods return `None`.

| Operation | Meaning |
| --- | --- |
| `a | b` | Union: in either group |
| `a & b` | Intersection: in both groups |
| `a - b` | Difference: in a but not b |
| `a ^ b` | Symmetric difference: in exactly one group |
| `a <= b` | Whether every item in a is also in b |

```python
trained = {1042, 2075, 4050}
trained.add(1042)
trained.add(5000)
trained.discard(5000)
trained.discard(9999)
required = {2075, 4050, 6000}
print("Trained:", sorted(trained))
print("Either group:", sorted(trained | required))
print("Both groups:", sorted(trained & required))
print("Still need training:", sorted(required - trained))
print("Trained but not required:", sorted(trained - required))
print("Exactly one group:", sorted(trained ^ required))
print("All required trained:", required <= trained)
```

## 8. Choose a collection

| Type | Main purpose | Mutable? | Duplicates? | Access |
| --- | --- | --- | --- | --- |
| `list` | Ordered collection to update | Yes | Yes | Position |
| `tuple` | Fixed ordered grouping | No | Yes | Position |
| `dict` | Named fields or key-based lookup | Yes | Keys unique | Key |
| `set` | Unique values and group comparisons | Yes | No | Membership |

Dictionary values may repeat. Strings, lists, and tuples are sequences; dictionaries and sets
serve different purposes. Choose based on the task rather than trying to use one type everywhere.
