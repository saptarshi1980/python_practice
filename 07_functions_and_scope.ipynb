## 1. Define and call a function

A function groups reusable work. Parameters name inputs; arguments supply values. `return` sends a result to the caller. Printing displays a value but does not return it. Define a function before calling it.

```python
def calculate_da(basic_pay):
    return basic_pay * 0.48

da = calculate_da(75_000)
print(f"DA: {da:,.2f}")
assert da == 36_000
```

## 2. Defaults and keyword arguments

Defaults make optional inputs explicit. Keyword arguments improve readability. A bare `*` makes following parameters keyword-only. Keep required inputs before optional positional inputs.

```python
def calculate_hra(basic_pay, *, rate=0.16, limit=16_000):
    return min(basic_pay * rate, limit)

print(calculate_hra(120_000))
print(calculate_hra(120_000, limit=18_000))
assert calculate_hra(120_000) == 16_000
```

## 3. Local scope and multiple results

Names assigned inside a function are local. Pass data in and return results instead of depending on changing global variables. A tuple return can be unpacked by the caller.

```python
def estimate_tax(annual_gross, threshold=1_200_000, rate=0.20):
    taxable = max(0, annual_gross - threshold)
    annual_tax = taxable * rate
    return annual_tax, annual_tax / 12

annual, monthly = estimate_tax(1_494_000)
print(annual, monthly)
assert (annual, monthly) == (58_800, 4_900)
```

## 4. Avoid mutable defaults

Default values are created when the function is defined. A list default would be shared between calls. Use `None` and create a new list inside the function when needed. A docstring describes the contract.

```python
def add_note(note, notes=None):
    """Return notes with one new note; create a fresh list when omitted."""
    if notes is None:
        notes = []
    notes.append(note)
    return notes

print(add_note("Review"))
print(add_note("Approved"))
assert add_note("A") == ["A"]
assert add_note("B") == ["B"]
```
