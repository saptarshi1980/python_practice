## 1. Catch specific conversion errors

Use `try/except` around an operation that can fail. Catch the expected exception rather than hiding every error. Validation messages belong at the user boundary.

```python
raw_pay = "not recorded"
try:
    basic_pay = float(raw_pay)
except ValueError as error:
    print("Enter a numeric salary.")
    print(type(error).__name__)
else:
    print(f"Accepted: {basic_pay:,.2f}")
```

## 2. Raise a useful error

A calculation function should reject inputs that violate its contract. `raise ValueError(...)` explains an invalid value. Floating-point NaN and infinity require explicit finite checks; neither is a usable salary.

```python
import math

def parse_pay(text):
    value = float(text.strip().replace(",", ""))
    if not math.isfinite(value) or value < 0:
        raise ValueError("Pay must be finite and nonnegative")
    return value

print(parse_pay(" 52,500.00 "))
for text in ["-1", "nan", "inf", "unknown"]:
    try:
        parse_pay(text)
    except ValueError:
        print("Rejected:", text)
```

## 3. Else, finally, and resources

`else` runs only if the try block succeeds; `finally` runs on exit, even after an exception. Prefer `with` for resources. Do not return from finally: it can suppress errors or override results.

```python
events = []
try:
    amount = int("42")
except ValueError:
    events.append("invalid")
else:
    events.append(f"accepted {amount}")
finally:
    events.append("finished")
print(events)
assert events == ["accepted 42", "finished"]
```

## 4. Keep processing valid rows

Handle an expected row error inside a loop when other rows can still be processed. Keep rejected rows and reasons rather than silently discarding them.

```python
valid = []
rejected = []
for row_number, text in enumerate(["50000", "bad", "-2", "60000"], start=1):
    try:
        valid.append(parse_pay(text))
    except ValueError as error:
        rejected.append({"row": row_number, "reason": str(error)})
print(valid)
print(rejected)
assert valid == [50_000, 60_000]
assert len(rejected) == 2
```
