## 1. List comprehensions

Use a comprehension for a short transformation or filter. Prefer a normal loop for multi-step logic, side effects, or complicated validation.

```python
basic_salaries = [40_000, -500, 50_000, 60_000]
valid_basic = [pay for pay in basic_salaries if pay >= 0]
da_values = [pay * 0.48 for pay in valid_basic]
print(valid_basic)
print(da_values)
assert da_values == [19_200, 24_000, 28_800]
```

## 2. Dictionary and set comprehensions

Dictionary comprehensions build key-value mappings; repeated keys overwrite previous values. Set comprehensions deduplicate transformed values. Normalize text deliberately before deduplicating.

```python
employees = [{"id": 101, "name": " Asha ", "dept": "Finance"},
             {"id": 102, "name": "Arun", "dept": " finance "}]
names_by_id = {item["id"]: item["name"].strip() for item in employees}
departments = {item["dept"].strip().casefold() for item in employees}
print(names_by_id)
print(sorted(departments))
assert departments == {"finance"}
```

## 3. Enumerate, zip, and mismatch checks

`enumerate` pairs items with counters. `zip` pairs collections and normally stops at the shortest. Use `strict=True` when mismatched lengths indicate bad data.

```python
names = ["Asha", "Arun"]
pay = [52_500, 75_000]
for number, name in enumerate(names, start=1):
    print(number, name)
records = [{"name": name, "basic": basic}
           for name, basic in zip(names, pay, strict=True)]
print(records)
try:
    list(zip(names, [52_500], strict=True))
except ValueError:
    print("Mismatched input lengths rejected.")
```

## 4. Unpacking and copying

Starred unpacking collects the remaining items into a list. Collection unpacking with `*` and `**` combines values; later dictionary keys win. A shallow copy still shares nested mutable objects.

```python
first, *middle, last = [10, 20, 30, 40]
print(first, middle, last)
base = {"electricity": 1000, "active": True}
custom = {**base, "electricity": 1500}
print(custom)
original = {"notes": ["Review"]}
copied = original.copy()
copied["notes"].append("Approved")
print(original)
assert original["notes"] == ["Review", "Approved"]
```
