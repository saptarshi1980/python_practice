## 1. A module is a Python file

A module groups reusable definitions. The temporary example imports a module by file location so it does not modify sys.path or leave project files behind. Import normally with `import payroll_math` when that file is in your project import path.

```python
import importlib.util
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    module_path = Path(folder) / "payroll_math.py"
    module_path.write_text("def calculate_da(basic):\n    return basic * 0.48\n", encoding="utf-8")
    spec = importlib.util.spec_from_file_location("lesson_payroll_math", module_path)
    payroll_math = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(payroll_math)
    print(payroll_math.calculate_da(75_000))
assert payroll_math.calculate_da(75_000) == 36_000
```

## 2. Import styles and namespaces

Prefer clear imports. `import module` keeps qualified names; `from module import name` brings selected names into the current namespace. Avoid wildcard imports and naming a file json.py, csv.py, or math.py because it can shadow a standard module.

```python
import math
from statistics import mean

print(math.isfinite(52_500))
print(mean([40_000, 50_000, 60_000]))
print(math.__name__)
assert mean([40_000, 50_000, 60_000]) == 50_000
```

## 3. The main guard

Imported code runs at import time. Keep demonstrations and command-line entry points under `if __name__ == "__main__":` so imports do not unexpectedly print or start processing files. Here exec simulates the two module names.

```python
source = """def greeting():
    return 'Payroll ready'

if __name__ == '__main__':
    print(greeting())
"""
namespace = {"__name__": "payroll_demo"}
exec(source, namespace)
print("Imported function:", namespace["greeting"]())
exec(source, {"__name__": "__main__"})
```

A small project can use payroll/calculations.py for formulas, payroll/io.py for reading and writing, and a main.py entry point. __init__.py marks a regular package. Keep dependencies one-way and avoid circular imports. Modules are cached in sys.modules; restart a notebook kernel when imports become stale during learning.

```python
project_layout = {
    "payroll/__init__.py": "Package entry",
    "payroll/calculations.py": "Pure salary functions",
    "payroll/io.py": "CSV and JSON boundaries",
    "main.py": "Validate inputs and coordinate work",
    "tests/test_calculations.py": "Calculation checks",
}
for path, purpose in project_layout.items():
    print(f"{path}: {purpose}")
```
