## 1. Iterable versus iterator

An iterable can provide an iterator. iter(...) creates an iterator; next(...) consumes one item. An exhausted iterator raises StopIteration, or returns a supplied default. A list can provide a fresh iterator each time.

```python
values = [10, 20]
iterator = iter(values)
print(next(iterator))
print(next(iterator))
print(next(iterator, "finished"))
print(list(iterator))
print(list(values))
assert list(iterator) == []
```

## 2. Generator functions

A function containing yield produces a generator iterator. Each yield pauses execution until the next item is requested. Generators are one-pass; create a new generator to repeat processing.

```python
def annual_gross_values(monthly_values):
    for monthly in monthly_values:
        yield monthly * 12

annual_values = annual_gross_values([87_100, 124_500])
print(list(annual_values))
print(list(annual_values))
assert list(annual_gross_values([100])) == [1200]
```

## 3. Generator expressions and pipelines

Parentheses can create a generator expression. Lazy pipelines avoid storing every intermediate result, though the source may still be a list. Laziness does not make errors disappear: they may occur during consumption.

```python
basic_values = [52_500, -1, 75_000]
valid_values = (basic for basic in basic_values if basic >= 0)
da_values = (basic * 0.48 for basic in valid_values)
total_da = sum(da_values)
print(total_da)
assert total_da == 61_200
assert list(da_values) == []
```

## 4. Stop after enough items

itertools.islice consumes a bounded portion. Put a file context around consumption, not only generator creation: a generator cannot read from a file that has already been closed.

```python
from itertools import islice

def numbered_records():
    number = 1
    while True:
        yield {"id": number}
        number += 1

first_three = list(islice(numbered_records(), 3))
print(first_three)
assert first_three == [{"id": 1}, {"id": 2}, {"id": 3}]
```
