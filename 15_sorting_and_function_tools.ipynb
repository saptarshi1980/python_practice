## 1. Functions as arguments

A function is a value that can be passed to another function. sorted(..., key=...) calls the key function for each record. It returns a new list and does not mutate the source.

```python
employees = [{"name": "Neha", "basic": 60_000},
             {"name": "Arun", "basic": 75_000},
             {"name": "Asha", "basic": 60_000}]

def salary_key(employee):
    return employee["basic"]

ordered = sorted(employees, key=salary_key)
print([row["name"] for row in ordered])
assert [row["name"] for row in employees] == ["Neha", "Arun", "Asha"]
```

## 2. Lambda and compound keys

A lambda is a short expression function. Use a named function when logic needs explanation. Tuple keys compare field by field. Python sorting is stable: equal keys keep their original order.

```python
ordered = sorted(employees, key=lambda row: (-row["basic"], row["name"].casefold()))
print([row["name"] for row in ordered])
assert [row["name"] for row in ordered] == ["Arun", "Asha", "Neha"]
print(max(employees, key=lambda row: row["basic"]))
```

## 3. Any and all

any(...) stops after the first true item; all(...) stops after the first false item. any([]) is False and all([]) is True. If a nonempty dataset is required, check it separately.

```python
print(any(row["basic"] < 0 for row in employees))
print(all(row["basic"] >= 0 for row in employees))
print(any([]), all([]))
records = []
valid_batch = bool(records) and all(row["basic"] >= 0 for row in records)
print("Nonempty valid batch:", valid_batch)
assert valid_batch is False
```

## 4. Partial application

functools.partial creates a callable with selected arguments preset. It can make policy-specific functions readable. Keep default rates visible rather than burying them in complex anonymous expressions.

```python
from functools import partial

def allowance(basic, *, rate):
    return basic * rate

calculate_da = partial(allowance, rate=0.48)
print(calculate_da(75_000))
assert calculate_da(75_000) == 36_000
```
